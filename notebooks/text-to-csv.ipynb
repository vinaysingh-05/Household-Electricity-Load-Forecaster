```python
import os
input_file = "../data/text data/household_power_consumption.txt"

# Read the text file
with open(input_file, "r", encoding="utf-8") as file:
    text_data = file.read()

# Convert semicolons to commas
csv_data = text_data.replace(';', ',')

# Create the folder and save
os.makedirs("../data/raw ", exist_ok=True)

with open("../data/raw/household_power.csv", "w", encoding="utf-8") as file:
    file.write(csv_data)
    
print("Saved successfully!")
```
